```python
# Exercise 3
n = int(input("Enter a number?"))
i = 2
while i < n :
    if n%i == 0:
        print(f"{n} is NOT a prime number")
        break
    i += 1
else:
    print(f"{n} is a prime number")
```

```python
# Exercise 4
n = int(input("Enter a number?"))
i = 2
divisor = [1]
while i < n:
    if n%i == 0:
        divisor.append(i)
    i += 1
if sum(divisor) == n:
    print(f"{n} is a perfect number")
else:
    print(f"{n} is NOT a perfect number")
```

```python
# Exercise 5
n = input("What is your favorite color? ")
colors = ["red", "green", "blue", "pink"]
i = 0
while i < len(colors) - 1:
    if n == colors[i]:
        print(f"Your color is at index {i} in my list")
        break
    i += 1
else:
    print("Sorry, i could not find your color")
```

```python
# Exercise 8
def extract_even(l):
    a = []
    for i in range(len(l)):
        if l[i]%2 == 0:
            a.append(l[i])
    return a
extract_even([1, 4, 5, -2, 10])
```

```python
# Exercise 9
def fac(n):
    i = 1
    a = 1
    if n == 0:
        return 1
    while i <= n:
        a *= i
        i += 1
    return a
fac (4)
```

```python
# Exercise 10
def get_divisors(n):
    i = 2
    divisor = [1]
    while i <= n:
        if n%i == 0:
            divisor.append(i)
        i += 1
    return divisor
get_divisors(10)
```

```python
# Exercise 11
import math
def point_distance(a, b):
    distance = math.sqrt((a[0]-b[0])**2 + (a[1] - b[1])**2)
    return distance
point_distance([1, 2], [5, 8])
```

```python
# Exercise 12
def print_dot(m, n):
    for x in range(1, m+1):
        if x == 1 or x == m:
            for z in range(1, n+1):
                if z == n:
                    print("*")
                else:
                    print("*", end=" ")

        else:
            for y in range(1, n+1):
                if y == 1:
                    print("*", end=" ")
                    for _ in range(1, n-2):
                        print(" ", end = "")
                if y == n:
                    print("*")
                else:
                    print(" ", end="")
                     
print_dot(10, 20)          
```
